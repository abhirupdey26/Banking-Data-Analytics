# Key Business Insights

## Customer & Loyalty Analysis
- The dataset contains 3,000 customers across different nationalities, loyalty classifications and fee structures.
- Jade is the largest loyalty segment with 1,331 customers.
- Jade customers also show the highest average deposits, loans and business lending among the loyalty segments.
- Silver customers have the highest average estimated income.

## Nationality Analysis
- European customers form the largest customer group with 1,309 customers.
- African customers have the highest average estimated income and average bank deposits.
- Australian customers have the highest average bank loans per customer.
- European customers contribute the highest total bank deposits because of their larger customer base.

## Fee Structure Analysis
- High-fee customers represent the largest fee segment with 1,476 customers.
- Jade is the dominant loyalty classification across all fee structures.
- The distribution of loyalty classifications is broadly similar across High, Mid and Low fee structures.

## Risk Analysis
- Risk Weighting ranges from 1 to 5, with Risk 2 being the largest group at 40.73% of customers.
- Approximately 68.60% of customers fall into Risk 1 or Risk 2.
- Higher-risk segments show higher average bank loans, deposits and credit-card balances.
- Risk 5 customers have the highest average bank loans (approximately ₹9.75 lakh) and average deposits (approximately ₹11.37 lakh).

## Financial Portfolio Analysis
- Business Lending represents the largest share of the combined financial value analyzed, at approximately 40.64%.
- Bank Deposits contribute approximately 31.49%.
- Bank Loans contribute approximately 27.73%.
- Credit Card Balance represents a much smaller share at approximately 0.15%.

## Overall Conclusion
The analysis shows that customer loyalty, nationality, fee structure and risk level are associated with different financial characteristics. Higher-risk customer segments in this dataset have greater average financial exposure, particularly through loans, deposits and credit-card balances. These insights can help a bank identify important customer segments and prioritize further analysis.
